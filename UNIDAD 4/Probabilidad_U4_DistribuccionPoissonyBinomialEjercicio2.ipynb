## Ejercicio 2 de distribuccion poisson y binomial

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

print("Tipo de distribución: Binomial (aproximada con Poisson)")
print("Porque hay muchos intentos (1000) y la probabilidad es muy pequeña (0.001)")

lam = 1  # λ = n * p = 1000 * 0.001

x = np.arange(0, 10)
px = poisson.pmf(x, lam)

p0 = poisson.pmf(0, lam)
p3 = poisson.pmf(3, lam)
p4 = poisson.pmf(4, lam)

print("\nP(X = 0) = {:.4f}".format(p0))
print("P(X = 3) = {:.4f}".format(p3))
print("P(X = 4) = {:.4f}".format(p4))
```

```python
plt.figure(figsize=(8,5))
plt.bar(x, px)
plt.title("Distribución de Poisson (λ = 1)")
plt.xlabel("Número de motores defectuosos")
plt.ylabel("Probabilidad")
plt.show()
```

```python
colors = ["blue"] * len(x)
colors[0] = "red"

plt.figure(figsize=(8,5))
plt.bar(x, px, color=colors)
plt.title("Probabilidad de que ninguno sea defectuoso")
plt.xlabel("Número de motores defectuosos")
plt.ylabel("Probabilidad")
plt.show()
```

```python
colors = ["blue"] * len(x)
colors[3] = "red"

plt.figure(figsize=(8,5))
plt.bar(x, px, color=colors)
plt.title("Probabilidad de que 3 sean defectuosos")
plt.xlabel("Número de motores defectuosos")
plt.ylabel("Probabilidad")
plt.show()
```
