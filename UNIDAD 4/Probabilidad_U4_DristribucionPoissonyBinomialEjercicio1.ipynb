## Ejercicio 1 de distribuccion de poisson y binomial.



```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import poisson

print("Tipo de distribución: Poisson")
print("Se usa porque se cuentan personas en un día y se conoce la media")

lam = 5

x = np.arange(0, 20)
px = poisson.pmf(x, lam)

p2 = poisson.pmf(2, lam)
p_mayor_10 = 1 - poisson.cdf(10, lam)

print("\nP(X = 2) = {:.4f}".format(p2))
print("P(X > 10) = {:.4f}".format(p_mayor_10))
```

```python
plt.figure(figsize=(8,5))
plt.bar(x, px)
plt.title("Distribución de Poisson")
plt.xlabel("Número de personas")
plt.ylabel("Probabilidad")
plt.show()
```

```python
colors = ["blue"] * len(x)
colors[2] = "red"

plt.figure(figsize=(8,5))
plt.bar(x, px, color=colors)
plt.title("Probabilidad de que sean 2 personas")
plt.xlabel("Número de personas")
plt.ylabel("Probabilidad")
plt.show()
```

```python
colors = ["blue" if i <= 10 else "red" for i in x]

plt.figure(figsize=(8,5))
plt.bar(x, px, color=colors)
plt.title("Probabilidad de que sean más de 10 personas")
plt.xlabel("Número de personas")
plt.ylabel("Probabilidad")
plt.text(6, 0.12, "P = {:.4f}".format(p_mayor_10))

plt.show()
```
