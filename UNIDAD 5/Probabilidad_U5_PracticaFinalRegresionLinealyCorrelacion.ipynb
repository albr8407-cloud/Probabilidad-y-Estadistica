```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
from scipy.stats import pearsonr
```

```python
import pandas as pd

grupo1 = pd.read_csv('grupo1.csv')
grupo2 = pd.read_csv('grupo2.csv')
grupo3 = pd.read_csv('grupo3.csv')
```

```python
print(grupo1.head())
print(grupo1.info())
print(grupo1.describe())
```

```python
plt.figure(figsize=(8,5))
plt.hist(grupo1['Calificacion'], bins=10)
plt.title('Histograma Grupo 1')
plt.xlabel('Calificación')
plt.ylabel('Frecuencia')
plt.grid(True)
plt.show()
```

```python
X1 = grupo1[['HorasEstudio']]
y1 = grupo1['Calificacion']

modelo1 = LinearRegression()
modelo1.fit(X1, y1)

y1_pred = modelo1.predict(X1)

pendiente1 = modelo1.coef_[0]
intercepto1 = modelo1.intercept_

print(f"y = {pendiente1:.4f}x + {intercepto1:.4f}")
```

```python
correlacion1, p1 = pearsonr(grupo1['HorasEstudio'], grupo1['Calificacion'])

r21 = r2_score(y1, y1_pred)

print('Correlación de Pearson:', correlacion1)
print('Valor p:', p1)
print('R²:', r21)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(grupo1['HorasEstudio'], grupo1['Calificacion'])
plt.plot(grupo1['HorasEstudio'], y1_pred, linewidth=2)

plt.title('Regresión Lineal Grupo 1')
plt.xlabel('Horas de estudio')
plt.ylabel('Calificación')
plt.grid(True)
plt.show()
```
